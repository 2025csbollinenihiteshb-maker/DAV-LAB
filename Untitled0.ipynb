```python
import pandas as pd
data = {
    "student_ID": [101,102,103,104,105,106,102,107],
    "student":["asha","ravi","priya","john",
               "anu","kiran","ravi","meena",],
    "department":["CSE","CSE","ECE","ECE",
                  "CSE","EEE","CSE","EEE"],
    "Attendance":[90,75,95,60,80,70,75,85],
    "internal_marks":[85,70,90,55,78,65,70,None],
    "Final_marks":[88,72,92,50,80,68,72,82]
}
```

```python
df = pd.DataFrame(data)
df.to_csv("students.csv", index=False)
print("students.csv created successfully!")
```

```python
df = pd.read_csv("students.csv")
print("original dataset:")
print(df)
```

```python
print("\nNumber of Duplicate Records:")
print(df.duplicated().sum())
df= df.drop_duplicates()
```

```python
print("\nAverage final marks by department:")
print(
    df.groupby("department")["Final_marks"].mean()
)
```
